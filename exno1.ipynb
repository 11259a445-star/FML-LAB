```python
import numpy as np

print(
    "1D Array:\n", np.array([1, 2, 3, 4, 5]), "\n\n",
    "2D Array:\n", np.array([[1, 2, 3], [4, 5, 6]]), "\n\n",
    "Zeros:\n", np.zeros((3, 3)), "\n\n",
    "Ones:\n", np.ones((2, 4)), "\n\n",
    "Identity:\n", np.eye(3), "\n\n",
    "Range:\n", np.arange(0, 10, 2)
)
```
