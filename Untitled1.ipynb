```python


import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Create dataset
data = {
    'StudyHours': [1, 2, 3, 4, 5, 6, 7, 8],
    'ExamScore': [35, 40, 50, 55, 65, 70, 78, 85]
}

df = pd.DataFrame(data)

# 1. Scatter Plot
plt.scatter(df['StudyHours'], df['ExamScore'])
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score (Scatter Plot)")
plt.show()

# 2. Line Plot
plt.plot(df['StudyHours'], df['ExamScore'], marker='o')
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Exam Score Progress (Line Plot)")
plt.show()

# 3. Histogram
plt.hist(df['ExamScore'], bins=2)
plt.xlabel("Score Range")
plt.ylabel("Number of Students")
plt.title("Exam Score Distribution (Histogram)")
plt.show()

# 4. Bar Chart
plt.bar(df['StudyHours'], df['ExamScore'])
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score (Bar Chart)")
plt.show()

```

```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# STEP 1: CREATE DATASET
data = {
    'Student': ['S1', 'S2', 'S3', 'S4', 'S5',
                'S6', 'S7', 'S8', 'S9', 'S10'],
    'Classes_Attended': [30, 35, 40, 45, 50,
                         55, 60, 65, 70, 75],
    'Internal_Marks': [35, 38, 42, 46, 50,
                       55, 60, 65, 70, 75]
}

df = pd.DataFrame(data)
print("Dataset:")
print(df)

# STEP 2: FEATURES & TARGET
X = df[['Classes_Attended']]
y = df['Internal_Marks']

# STEP 3: TRAIN-TEST SPLIT
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=7
)

# STEP 4: TRAIN MODEL
model = LinearRegression()
model.fit(X_train, y_train)

# STEP 5: PREDICTION
predictions = model.predict(X_test)

print("\nActual Marks:", y_test.tolist())
print("Predicted Marks:", predictions)

# STEP 6: EVALUATION
mse = mean_squared_error(y_test, predictions)

print("\nMean Squared Error (MSE):", mse)
print("Marks per Class:", model.coef_[0])
print("Base Marks:", model.intercept_)

# STEP 7: VISUALIZATION
plt.scatter(
    X['Classes_Attended'],
    y,
    color='blue',
    label='Actual Data'
)

plt.plot(
    X['Classes_Attended'],
    model.predict(X),
    color='red',
    label='Regression Line'
)

plt.xlabel("Classes Attended")
plt.ylabel("Internal Marks")
plt.title("Classes Attended vs Internal Marks")
plt.legend()
plt.show()
```
